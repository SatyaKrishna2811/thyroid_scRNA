# Phase 8: In Silico Knockdown Perturbation & Pseudotime Therapy Reversal

### Overview
In this analysis, we evaluate the therapeutic potential of candidate drivers identified in single-cell trajectory analysis to induce redifferentiation in radioiodine-refractory thyroid carcinoma.

Specifically, we test:
1. **Driver Candidates**: Key EMT, metastatic, and dedifferentiation genes anti-correlated with thyroid avidity (`S100A6`, `FN1`, `S100A4`, `LGALS3`).
2. **Transcription Factor Candidates**: AP-1 complex members (`JUN`, `FOS`) alongside hypothesized regulators (`RUNX2`, `SOX9`).
3. **Differentiation Controls**: Lineage-defining master factors (`PAX8`, `NKX2-1`).
4. **Random Gene Controls**: High-expression non-differentiation control genes (`ACTB`, `GAPDH`, `B2M`, `EEF1A1`, `MALAT1`).

For each gene, we quantify:
- **Delta RAI Index**: Predicted shift in the 10-gene RAI avidity score upon gene inactivation.
- **Shift in Pseudotime**: Trajectory repositioning towards differentiated Paratumor state (negative shift indicates therapeutic reversal).


```
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

PROCESSED_DIR = os.path.join(os.getcwd(), "processed_data")

res_df = pd.read_csv(os.path.join(PROCESSED_DIR, "in_silico_knockdown_results.csv"))
print(f"Loaded in silico knockdown results for {len(res_df)} genes.")
res_df
```

### Cell 2: In Silico Knockdown Visualization
We plot the predicted change in RAI Index (left) and shift along diffusion pseudotime (right).


```
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

colors = {
    "Driver Candidate": "#d62728",
    "TF Candidate": "#ff7f0e",
    "Differentiation Factor": "#2ca02c",
    "Random Control": "#7f7f7f"
}

bar_colors = [colors[t] for t in res_df["Type"]]
y_pos = np.arange(len(res_df))

# Subplot 1: Delta RAI Index
axes[0].barh(y_pos, res_df["Delta_RAI_Index"], color=bar_colors, alpha=0.85)
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(res_df["Gene"], fontsize=11)
axes[0].invert_yaxis()
axes[0].axvline(0, color="black", linestyle="--", linewidth=0.8)
axes[0].set_xlabel("Predicted Delta RAI Index (Redifferentiation > 0)", fontsize=12)
axes[0].set_title("In Silico Knockdown Effect on RAI Index", fontsize=13, fontweight="bold")

# Subplot 2: Shift in Pseudotime
axes[1].barh(y_pos, res_df["Shift_in_Pseudotime"], color=bar_colors, alpha=0.85)
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(res_df["Gene"], fontsize=11)
axes[1].invert_yaxis()
axes[1].axvline(0, color="black", linestyle="--", linewidth=0.8)
axes[1].set_xlabel("Predicted Shift in Pseudotime (Reversal toward Paratumor < 0)", fontsize=12)
axes[1].set_title("In Silico Knockdown Trajectory Reversal", fontsize=13, fontweight="bold")

legend_elements = [Patch(facecolor=c, label=l) for l, c in colors.items()]
axes[1].legend(handles=legend_elements, loc="lower right", frameon=True)

plt.tight_layout()
plt.show()
```

### Cell 3: Mechanistic Summary of Candidate Drivers vs Hypothesized Regulators
- **AP-1 complex (JUN/FOS) and S100/FN1 drivers**: Knockdown reliably restores thyroid-specific gene expression and yields negative pseudotime shifts, indicating reversal toward differentiated avidity.
- **RUNX2 & SOX9 evaluation**: Demonstrates low expression in thyrocytes (6.3% and 16.3% positivity) with negligible predicted impact (Delta RAI ~ +0.0036 and -0.0135), supporting their classification as candidate drivers rather than master repressors.
- **PAX8 & NKX2-1**: Inactivation precipitates marked loss of differentiation (negative Delta RAI and positive pseudotime shift), confirming model fidelity.
